```python
import pandas as pd

ball = pd.read_csv(r"C:\Users\ckcha\OneDrive\Desktop\CSK_Dashboard\Data\ball_by_ball_data.csv")
matches = pd.read_csv(r"C:\Users\ckcha\OneDrive\Desktop\CSK_Dashboard\Data\ipl_matches_data.csv")
players = pd.read_csv(r"C:\Users\ckcha\OneDrive\Desktop\CSK_Dashboard\Data\players-data-updated.csv")
teams = pd.read_csv(r"C:\Users\ckcha\OneDrive\Desktop\CSK_Dashboard\Data\teams_data.csv")

CSK_ID = 129

csk_matches = matches[
    (matches["team1"] == CSK_ID) |
    (matches["team2"] == CSK_ID)
]

print(csk_matches.shape)

csk_matches.head()

csk_matches["season"].value_counts().sort_index()


csk_matches.to_csv(
    r"C:\Users\ckcha\OneDrive\Desktop\CSK_Dashboard\Clean_Data\csk_matches.csv",
    index=False
)

print(csk_matches.shape)

csk_matches["season"].value_counts().sort_index()


#Create Win Flag
csk_matches["csk_win"] = (
    csk_matches["match_winner"] == CSK_ID
).astype(int)


#Create Season Summary

csk_season_summary = csk_matches.groupby(
    "season"
).agg(
    matches=("match_id","count"),
    wins=("csk_win","sum")
).reset_index()

#Losses

csk_season_summary["losses"] = (
    csk_season_summary["matches"]
    -
    csk_season_summary["wins"]
)

#Win Percentage

csk_season_summary["win_percentage"] = (
    csk_season_summary["wins"]
    /
    csk_season_summary["matches"]
    * 100
).round(2)

csk_season_summary

csk_season_summary.to_csv(
    r"C:\Users\ckcha\OneDrive\Desktop\CSK_Dashboard\Clean_Data\csk_season_summary.csv",
    index=False
)

csk_season_summary.head()

csk_season_summary.tail()
```

```python
#Dashboard Page 2
#CSK Match IDs
csk_match_ids = csk_matches["match_id"].unique()

clean_ball = pd.read_csv(
    r"C:\Users\ckcha\OneDrive\Desktop\CSK_Dashboard\Data\clean_ball_data.csv"
)

csk_ball = clean_ball[
    clean_ball["match_id"].isin(csk_match_ids)
]


csk_batting = csk_ball[
    csk_ball["team_batting"] == CSK_ID
]

csk_batting_analysis = csk_batting.groupby(
    "batter"
).agg(
    runs=("batter_runs","sum"),
    balls=("ball_number","count"),
    boundaries=("boundary_flag","sum")
).reset_index()

csk_batting_analysis["strike_rate"] = (
    csk_batting_analysis["runs"]
    /
    csk_batting_analysis["balls"]
    * 100
).round(2)

csk_batting_analysis = csk_batting_analysis.sort_values(
    by="runs",
    ascending=False
)

csk_batting_analysis.head(20)

csk_batting_analysis.to_csv(
    r"C:\Users\ckcha\OneDrive\Desktop\CSK_Dashboard\Clean_Data\csk_batting_analysis.csv",
    index=False
)
```

```python
#Filter CSK Bowling

csk_bowling = csk_ball[
    csk_ball["team_bowling"] == CSK_ID
]

#Create Bowling Summary

csk_bowling_analysis = csk_bowling.groupby(
    "bowler"
).agg(
    wickets=("wicket_flag","sum"),
    balls=("ball_number","count"),
    dot_balls=("dot_ball_flag","sum"),
    runs_conceded=("total_runs","sum")
).reset_index()

#Economy

csk_bowling_analysis["overs"] = (
    csk_bowling_analysis["balls"] / 6
)

csk_bowling_analysis["economy"] = (
    csk_bowling_analysis["runs_conceded"]
    /
    csk_bowling_analysis["overs"]
).round(2)

#Sort
csk_bowling_analysis = csk_bowling_analysis.sort_values(
    by="wickets",
    ascending=False
)


csk_bowling_analysis.head(20)

csk_bowling_analysis.to_csv(
    r"C:\Users\ckcha\OneDrive\Desktop\CSK_Dashboard\Clean_Data\csk_bowling_analysis.csv",
    index=False
)
```

```python
#Create venue summary

csk_matches["venue"] = csk_matches["venue"].replace({
    "MA Chidambaram Stadium": "MA Chidambaram Stadium, Chepauk",
    "MA Chidambaram Stadium, Chepauk, Chennai": "MA Chidambaram Stadium, Chepauk",
    "Wankhede Stadium": "Wankhede Stadium, Mumbai",
    "Dr DY Patil Sports Academy": "Dr DY Patil Sports Academy, Mumbai",
    "Feroz Shah Kotla": "Arun Jaitley Stadium"
})

csk_venue_analysis = csk_matches.groupby(
    "venue"
).agg(
    matches=("match_id","count"),
    wins=("csk_win","sum")
).reset_index()


#Losses

csk_venue_analysis["losses"] = (
    csk_venue_analysis["matches"]
    -
    csk_venue_analysis["wins"]
)

#Win %

csk_venue_analysis["win_percentage"] = (
    csk_venue_analysis["wins"]
    /
    csk_venue_analysis["matches"]
    * 100
).round(2)

#Sort

csk_venue_analysis = csk_venue_analysis.sort_values(
    by="win_percentage",
    ascending=False
)


csk_venue_analysis = csk_venue_analysis[
    csk_venue_analysis["matches"] >= 5
]

csk_venue_analysis = csk_venue_analysis.sort_values(
    by="win_percentage",
    ascending=False
)

csk_venue_analysis.head(15)


csk_venue_analysis.to_csv(
    r"C:\Users\ckcha\OneDrive\Desktop\CSK_Dashboard\Clean_Data\csk_venue_analysis.csv",
    index=False
)
```

```python
#Opponent Analysis

csk_matches["opponent"] = csk_matches.apply(
    lambda row:
    row["team2"] if row["team1"] == CSK_ID
    else row["team1"],
    axis=1
)

csk_opponent_analysis = csk_matches.groupby(
    "opponent"
).agg(
    matches=("match_id","count"),
    wins=("csk_win","sum")
).reset_index()

csk_opponent_analysis["losses"] = (
    csk_opponent_analysis["matches"]
    -
    csk_opponent_analysis["wins"]
)

csk_opponent_analysis["win_percentage"] = (
    csk_opponent_analysis["wins"]
    /
    csk_opponent_analysis["matches"]
    * 100
).round(2)

csk_opponent_analysis = csk_opponent_analysis.sort_values(
    by="win_percentage",
    ascending=False
)

team_lookup = dict(
    zip(teams["team_id"], teams["team_name"])
)

csk_opponent_analysis["opponent_name"] = (
    csk_opponent_analysis["opponent"]
    .map(team_lookup)
)

csk_opponent_analysis[
    [
        "opponent_name",
        "matches",
        "wins",
        "losses",
        "win_percentage"
    ]
]

csk_opponent_analysis.to_csv(
    r"C:\Users\ckcha\OneDrive\Desktop\CSK_Dashboard\Clean_Data\csk_opponent_analysis.csv",
    index=False
)
```
